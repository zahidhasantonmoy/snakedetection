```python
import os
import shutil
import random
from PIL import Image  # For checking corruption and getting size
import warnings

# Suppress warnings for cleaner output
warnings.filterwarnings("ignore")

# Paths - customize if needed
original_root = '/kaggle/input/snake-6-gb/final_processed_dataset/'
new_root = '/kaggle/working/new-balanced-dataset/'
target_count = 100

# Create new root if it doesn't exist
os.makedirs(new_root, exist_ok=True)

# Function to check if image is valid and get its quality metric (resolution: width * height)
def get_image_quality(image_path):
    try:
        with Image.open(image_path) as img:
            img.verify()  # Verify if corrupted
            img = Image.open(image_path)  # Re-open to get size (verify closes it)
            size = img.width * img.height
            return size
    except (IOError, SyntaxError, ValueError):
        return -1  # Invalid or corrupted

# Process each species folder
for species in os.listdir(original_root):
    original_folder = os.path.join(original_root, species)
    if not os.path.isdir(original_folder):
        continue  # Skip non-folders
    
    # Get list of image files
    images = [f for f in os.listdir(original_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    valid_images = []
    
    print(f"Processing {species} with {len(images)} images...")
    
    # Check each image for corruption and get quality
    for img_file in images:
        img_path = os.path.join(original_folder, img_file)
        quality = get_image_quality(img_path)
        if quality > 0:
            valid_images.append((img_path, quality))
        else:
            print(f"Skipping corrupted/low-quality image: {img_path}")
    
    num_valid = len(valid_images)
    print(f"{species} has {num_valid} valid images.")
    
    # Create new species folder
    new_folder = os.path.join(new_root, species)
    os.makedirs(new_folder, exist_ok=True)
    
    if num_valid >= target_count:
        # Sort by quality descending (best first)
        valid_images.sort(key=lambda x: x[1], reverse=True)
        # Select top 100 best quality
        selected = [img[0] for img in valid_images[:target_count]]
    else:
        # Keep all valid if fewer than 100 (already sorted if you want, but not necessary)
        valid_images.sort(key=lambda x: x[1], reverse=True)
        selected = [img[0] for img in valid_images]
        print(f"Warning: {species} has only {num_valid} valid images, keeping all.")
    
    # Copy selected images
    for src in selected:
        dst = os.path.join(new_folder, os.path.basename(src))
        shutil.copy(src, dst)

print("New dataset created successfully at", new_root)
print("Run this in your Kaggle notebook. If PIL is not installed, add '!pip install pillow' at the top.")
```
